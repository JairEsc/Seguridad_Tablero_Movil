```python
import requests
from bs4 import BeautifulSoup
import boto3
from datetime import datetime
```

```python
url = "https://www.gob.mx/sesnsp/acciones-y-programas/datos-abiertos-de-incidencia-delictiva"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
ultima_actualizacion='Secretariado Ejecutivo del Sistema Nacional de Seguridad Pública |\n\n            20 de febrero de 2026'


fecha_ultima_actualizacion = soup.find("section", {"class": "-box"}).find("p").text
    

if((ultima_actualizacion) not in fecha_ultima_actualizacion):
    print("Se actualizó")
    def enviar_email(fecha):
        ses = boto3.client('ses', region_name='us-east-2')
        ses.send_email(
            Source='eugenio.escobar@hidalgo.gob.mx',
            Destination={'ToAddresses': ['eugenio.escobar@hidalgo.gob.mx']},
            Message={
                'Subject': {'Data': 'Ya puedes actualizar el tablero de seguridad'},
                'Body': {'Text': {'Data': f'La base de datos se actualizó el {fecha}. Ya puedes descargarla.'}}
            }
        )
    
    enviar_email(fecha_ultima_actualizacion)
```
